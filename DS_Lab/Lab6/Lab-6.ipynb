# Part 1 — Quick start: will a person buy life insurance?

📘 **Problem.** We want to predict whether a person **buys life insurance (1)** or **does not (0)** from their **age**.

This is a **binary classification** problem: the output has only two possible values, $y \in \{0, 1\}$.

**Why not plain linear regression?** A straight line $y = mx + b$ can output values like $-0.4$ or $1.7$, which make no sense as a "yes/no" answer or as a probability. Logistic regression keeps the linear part but **squashes** it into the range $(0,1)$ with the **sigmoid** function, so the output can be read as a probability:

$$
h_\theta(x) = P(y = 1 \mid x;\theta) = \sigma(\theta^T x) = \frac{1}{1+e^{-\theta^T x}}
$$

The whole model at a glance (from the lecture slides):

**Read the diagram left → right:**
1. Inputs $x_1,\dots,x_n$ (plus a constant $x_0 = 1$ for the bias) are multiplied by weights $\theta_0,\dots,\theta_n$.
2. **Linear part (logit):** $z = \theta^T x = \theta_0 + \theta_1 x_1 + \dots + \theta_n x_n$
3. **Sigmoid activation:** $h_\theta(x) = \sigma(z)$ — a number between 0 and 1.
4. **Decision rule:** predict class **1** if $h_\theta(x) \ge 0.5$, otherwise class **0**.

# The loss and cost functions

### Binary Cross-Entropy (log loss) for ONE example
The **Binary Cross-Entropy Loss** (also called **log loss**) measures how wrong one predicted probability is. With $h = h_\theta(x)$ and true label $y \in \{0,1\}$:

$$ \boxed{L\big(h_\theta(x), y\big) = -\Big(y \log h_\theta(x) + (1-y)\log\big(1-h_\theta(x)\big)\Big)} $$

Because $y$ is either 0 or 1, only **one** of the two terms is ever "switched on":

| True label | Loss becomes | What the model must do to make the loss small |
|---|---|---|
| $y = 1$ | $L = -\log h_\theta(x)$ | make $h_\theta(x)$ **large** (close to 1) |
| $y = 0$ | $L = -\log\big(1-h_\theta(x)\big)$ | make $h_\theta(x)$ **small** (close to 0) |

💡 $\log$ here is the **natural log** (`np.log`). A confident **wrong** prediction is punished very heavily: as $h \to 0$ when $y=1$, the loss $\to \infty$.

# Prepare the X and Y

```python
X = df[['Height', 'Weight']].values
y = df['Gender'].values.reshape(-1, 1)
# Normalize features
X = (X - X.mean(axis=0)) / X.std(axis=0)

# Add bias term (intercept)
X = np.hstack([np.ones((X.shape[0], 1)), X])  # shape (n, 3)
print(X.shape)
print(y.shape)

print(X)
```

# The cost function for the WHOLE dataset
The **cost** $J(\theta)$ is simply the **average loss** over all $m$ training examples:

$$ \boxed{J(\theta) = -\frac{1}{m}\sum_{i=1}^{m}\Big[\,y^{(i)}\log h_\theta(x^{(i)}) + \big(1-y^{(i)}\big)\log\big(1-h_\theta(x^{(i)})\big)\Big]} $$

**Notation used in this lab**

| Symbol | Meaning |
|---|---|
| $J(\theta)$ | Cost function (depends on the parameters $\theta$) |
| $m$ | Number of training examples |
| $n$ | Number of features |
| $x^{(i)}$ | Feature vector of the $i$-th example (with $x_0^{(i)} = 1$) |
| $y^{(i)}$ | True label of the $i$-th example (0 or 1) |
| $\theta$ | Parameter vector $[\theta_0, \theta_1, \dots, \theta_n]^T$ ($\theta_0$ = bias) |
| $z^{(i)} = \theta^T x^{(i)}$ | Linear score (logit) of the $i$-th example |
| $h_\theta(x^{(i)}) = \sigma(z^{(i)})$ | Predicted probability for the $i$-th example |
| $\sigma(\cdot)$ | Sigmoid activation function |
| $\alpha$ | Learning rate |

🎯 **Goal of training:** find the $\theta$ that makes $J(\theta)$ as small as possible. For logistic regression, $J(\theta)$ is **convex** (bowl-shaped), so gradient descent can reach the global minimum.

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def compute_loss(y, y_pred):
    m = len(y)
    epsilon = 1e-8  # avoid log(0)
    return - (1/m) * np.sum(y * np.log(y_pred + epsilon) + (1-y) * np.log(1 - y_pred + epsilon))
```

# Deriving the gradient with the chain rule

📘 To minimise $J(\theta)$ with gradient descent we need $\dfrac{\partial L}{\partial \theta_j}$ — how the loss changes when we nudge one parameter. The loss depends on $\theta_j$ **through a chain**:

$$ \theta_j \;\longrightarrow\; z = \theta^T x \;\longrightarrow\; h_\theta(x) = \sigma(z) \;\longrightarrow\; L(h_\theta(x), y) $$

So by the **chain rule**:

$$ \frac{\partial L}{\partial \theta_j} = \frac{\partial L}{\partial h_\theta(x)} \cdot \frac{\partial h_\theta(x)}{\partial z} \cdot \frac{\partial z}{\partial \theta_j} $$



# Algorithm: Gradient Descent for Logistic Regression


\begin{array}{ll}
\textbf{Input:} & X,\ y,\ \alpha=0.01,\ T=1000\\
\textbf{Output:} & \theta,\ \text{loss\_history}
\end{array}


\begin{aligned}
&\textbf{Begin}\\
&\quad m \gets \text{number of samples in } X\\
&\quad n \gets \text{number of features in } X\\
&\quad \theta \gets \mathbf{0}_{n\times 1}\\
&\quad \text{loss\_history} \gets [\,]\\[2pt]
&\quad \textbf{For } t=1 \textbf{ to } T \textbf{ do}\\
&\qquad z \gets X\theta\\
&\qquad \hat{y} \gets \sigma(z)
       = \frac{1}{1+e^{-z}}\\
&\qquad e \gets \hat{y}-y\\
&\qquad g \gets \frac{1}{m}X^{\top}e\\
&\qquad \theta \gets \theta-\alpha g\\
&\qquad L \gets \operatorname{ComputeLoss}(y,\hat{y})\\
&\qquad \text{Append } L \text{ to loss\_history}\\
&\quad \textbf{End For}\\[2pt]
&\quad \textbf{Return } \theta,\ \text{loss\_history}\\
&\textbf{End}
\end{aligned}



```python
#T is the number of epochs
def gradient_descent(X, y, lr=0.1, epochs=1000):
	m=X.shape[0] #row
	n=X.shape[1] #column
	theta = np.zeros((n,1))
	L = []
	for i in range(epochs):
		z = np.dot(X,theta)
		y_pred = sigmoid(z)
		e = y_pred - y
		grad = (1/m) * np.dot(X.T,e)
		theta -= lr * grad
		loss = compute_loss(y,y_pred)
		L.append(loss)
	return theta,L
```

# ---------------------------
# Step 5: Train Model
# ---------------------------

```python

theta, loss_history = gradient_descent(X, y, lr=0.1, epochs=1000)

# ---------------------------
# Step 6: Predictions
# ---------------------------
y_pred_prob = sigmoid(np.dot(X, theta))
y_pred = (y_pred_prob >= 0.5).astype(int)

# ---------------------------
# Step 7: Plot Loss Curve
# ---------------------------
plt.plot(loss_history)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Loss Curve")
plt.show()
```

### Precision, Recall, F1-score and Accuracy

| Metric | Formula | Question it answers |
|---|---|---|
| **Accuracy** | $\dfrac{TP + TN}{TP + TN + FP + FN}$ | What fraction of *all* predictions are correct? (overall correctness) |
| **Precision** | $\dfrac{TP}{TP + FP}$ | Of everything I *predicted* positive, how much really is positive? (accuracy of the positive predictions) |
| **Recall** (Sensitivity, TPR) | $\dfrac{TP}{TP + FN}$ | Of all *actual* positives, how many did I catch? |
| **F1-score** | $2 \times \dfrac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}$ | Harmonic mean — balances precision and recall; useful when classes are **imbalanced** |

💡 **When to care about which?**
- **High recall** matters when *missing* a positive is costly (disease screening, fraud detection).
- **High precision** matters when a *false alarm* is costly (spam filter deleting real emails).

🔗 More: https://developers.google.com/machine-learning/crash-course/classification/accuracy-precision-recall

```python
# ---------------------------
# Step 8: Confusion Matrix
# ---------------------------
cm = confusion_matrix(y, y_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Female", "Male"],
            yticklabels=["Female", "Male"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix")
plt.show()

# ---------------------------
# Step 9: Classification Report
# ---------------------------
print(classification_report(y, y_pred, target_names=["Female", "Male"]))
```

### 1. Load and inspect the dataset
Check the shape, column names, sample rows, data types, missing values and duplicate records.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report, confusion_matrix, roc_auc_score, RocCurveDisplay
)

sns.set_theme(style="whitegrid", context="notebook")
df = pd.read_csv("HR_Analytics.csv")

display(df.head())
print("Dataset shape:", df.shape)
display(df.dtypes.rename("Data type").to_frame())
display(df.isnull().sum().rename("Missing values").to_frame())
print("Duplicate rows:", df.duplicated().sum())
display(df.describe().T)
```

### 2. Explore employee retention
First examine the target variable. Then compare numeric features for employees who stayed and left. Differences are associations, not proof of causation.

```python
target_labels = df["left"].map({0: "Stayed", 1: "Left"})
display(pd.DataFrame({
    "Employee count": df["left"].value_counts().sort_index(),
    "Percentage (%)": (df["left"].value_counts(normalize=True).sort_index() * 100).round(2)
}).rename(index={0: "Stayed (0)", 1: "Left (1)"}))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(data=df, x="left", hue="left", legend=False, ax=axes[0], palette="Set2")
axes[0].set(title="Employee retention outcome", xlabel="Left (0 = stayed, 1 = left)", ylabel="Number of employees")
axes[0].set_xticks([0, 1], ["Stayed", "Left"])
(df["left"].value_counts(normalize=True).sort_index() * 100).reindex([0, 1]).plot(
    kind="pie", autopct="%1.1f%%", startangle=90, ax=axes[1], labels=["Stayed", "Left"]
)
axes[1].set_ylabel("")
axes[1].set_title("Share of employees")
plt.tight_layout()
plt.show()

numeric_cols = [
    "satisfaction_level", "last_evaluation", "number_project",
    "average_montly_hours", "time_spend_company", "Work_accident",
    "promotion_last_5years"
]
display(df.groupby("left")[numeric_cols].mean().T.rename(columns={0: "Stayed", 1: "Left"}).round(3))

fig, axes = plt.subplots(3, 3, figsize=(15, 11))
axes = axes.flatten()
for i, col in enumerate(numeric_cols):
    sns.boxplot(data=df, x="left", y=col, hue="left", legend=False, ax=axes[i], palette="Set2")
    axes[i].set_title(col.replace("_", " ").title())
    axes[i].set_xlabel("")
    axes[i].set_xticks([0, 1], ["Stayed", "Left"])
for ax in axes[len(numeric_cols):]:
    ax.remove()
plt.suptitle("Numeric features by employee outcome", y=1.02, fontsize=15)
plt.tight_layout()
plt.show()
```

**Interpretation:** Compare the group means and boxplots to identify features that differ between employees who stayed and left. Satisfaction, project load, monthly hours and tenure are useful candidates to inspect. These patterns alone do not establish causation.

### 3. Salary level and retention
Compare both employee counts and attrition rates. Rates are more suitable than raw counts for comparing salary groups of different sizes.

```python
salary_order = ["low", "medium", "high"]
salary_counts = pd.crosstab(df["salary"], df["left"]).reindex(salary_order, fill_value=0)
salary_rates = (df.groupby("salary", observed=False)["left"].mean() * 100).reindex(salary_order)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
salary_counts.rename(columns={0: "Stayed", 1: "Left"}).plot(
    kind="bar", stacked=True, ax=axes[0], color=["#70AD91", "#D97979"]
)
axes[0].set(title="Employee count by salary", xlabel="Salary level", ylabel="Employees")
axes[0].tick_params(axis="x", rotation=0)
sns.barplot(x=salary_rates.index, y=salary_rates.values, ax=axes[1], color="#6C91BF")
axes[1].set(title="Attrition rate by salary", xlabel="Salary level", ylabel="Employees who left (%)")
axes[1].set_ylim(0, max(salary_rates.max() * 1.2, 1))
for container in axes[1].containers:
    axes[1].bar_label(container, fmt="%.1f%%")
plt.tight_layout()
plt.show()

display(pd.DataFrame({
    "Employees": df["salary"].value_counts().reindex(salary_order),
    "Attrition rate (%)": salary_rates.round(2)
}))
```

**Interpretation:** Use the attrition-rate chart to compare the proportion leaving at each salary level; the stacked chart shows the corresponding group sizes.

### 4. Department and retention
Departments differ in size, so inspect both counts and attrition rates.

```python
department_counts = pd.crosstab(df["Department"], df["left"])
department_rates = (df.groupby("Department")["left"].mean() * 100).sort_values(ascending=False)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
department_counts.rename(columns={0: "Stayed", 1: "Left"}).sort_index().plot(
    kind="bar", stacked=True, ax=axes[0], color=["#70AD91", "#D97979"]
)
axes[0].set(title="Employee count by department", xlabel="Department", ylabel="Employees")
axes[0].tick_params(axis="x", rotation=35)
sns.barplot(x=department_rates.values, y=department_rates.index, ax=axes[1], color="#6C91BF")
axes[1].set(title="Attrition rate by department", xlabel="Employees who left (%)", ylabel="Department")
for container in axes[1].containers:
    axes[1].bar_label(container, fmt="%.1f%%", padding=3)
plt.tight_layout()
plt.show()

display(pd.DataFrame({
    "Employees": df["Department"].value_counts(),
    "Attrition rate (%)": (df.groupby("Department")["left"].mean() * 100).round(2)
}).sort_values("Attrition rate (%)", ascending=False))
```

### 5. Correlation and feature selection
The correlation heatmap is a quick screen for numeric variables. Logistic regression can also use categorical features after one-hot encoding. Based on the EDA, use the numeric employee attributes plus salary and department.

```python
plt.figure(figsize=(10, 7))
corr = df[numeric_cols + ["left"]].corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, linewidths=.4)
plt.title("Correlation matrix of numeric features")
plt.tight_layout()
plt.show()

features = numeric_cols + ["salary", "Department"]
X = df[features].copy()
y = df["left"].copy()
print("Selected features:", features)
print("Target: left (1 = employee left, 0 = employee stayed)")
```

### 6. Split the data and train logistic regression
Use an 80:20 split with stratification to preserve the outcome proportions. Scale numeric features and one-hot encode categorical features within a pipeline, preventing preprocessing from being fitted on the test set.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

categorical_features = ["salary", "Department"]
preprocessor = ColumnTransformer([
    ("numeric", StandardScaler(), numeric_cols),
    ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_features)
])

model = Pipeline([
    ("preprocessor", preprocessor),
    ("logistic_regression", LogisticRegression(max_iter=1000, class_weight=None))
])
model.fit(X_train, y_train)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
print("Model training complete.")
```

### 7. Predictions and model evaluation
Accuracy is the fraction classified correctly. Precision, recall and F1-score for class `1` are especially important because they describe how well the model identifies employees who leave.

```python
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]

metrics = pd.DataFrame({
    "Metric": ["Accuracy", "Precision (left)", "Recall (left)", "F1-score (left)", "ROC-AUC"],
    "Score": [
        accuracy_score(y_test, y_pred),
        precision_score(y_test, y_pred, zero_division=0),
        recall_score(y_test, y_pred, zero_division=0),
        f1_score(y_test, y_pred, zero_division=0),
        roc_auc_score(y_test, y_prob)
    ]
})
display(metrics.round(4))
print(classification_report(y_test, y_pred, target_names=["Stayed", "Left"], zero_division=0))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
cm = confusion_matrix(y_test, y_pred, labels=[0, 1])
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
            xticklabels=["Stayed", "Left"], yticklabels=["Stayed", "Left"], ax=axes[0])
axes[0].set(title="Confusion matrix", xlabel="Predicted outcome", ylabel="Actual outcome")
RocCurveDisplay.from_predictions(y_test, y_prob, ax=axes[1], name="Logistic regression")
axes[1].plot([0, 1], [0, 1], linestyle="--", color="gray", label="Random classifier")
axes[1].set_title("ROC curve")
axes[1].legend()
plt.tight_layout()
plt.show()
```

**Confusion matrix:** top-left = correctly predicted stayed; top-right = stayed but predicted left; bottom-left = left but predicted stayed; bottom-right = correctly predicted left.
